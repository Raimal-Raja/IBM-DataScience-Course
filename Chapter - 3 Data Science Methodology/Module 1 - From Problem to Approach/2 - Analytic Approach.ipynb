# Data Science Methodology 101: From Problem to Approach – Analytic Approach

## Introduction

After establishing a clear business understanding, the next step in the data science methodology is selecting the appropriate analytic approach. The success of a data science project depends heavily on choosing an analytical method that aligns with the business question and project requirements. Different questions require different approaches, and selecting the right one ensures that the analysis produces meaningful and actionable results.

## Understanding the Analytic Approach

The analytic approach phase focuses on determining how the problem will be solved using data.

Once the business problem has been clearly defined, data scientists identify the type of analysis required to answer the question effectively. This involves understanding the nature of the problem and selecting techniques capable of uncovering the necessary insights.

The chosen approach should directly support the business objectives established during the Business Understanding phase.

## Matching Questions to Analytical Techniques

Different types of business questions require different analytical approaches.

### Predictive Analysis

When the goal is to estimate the likelihood of future events or outcomes, predictive models are often used.

Examples include:

* Predicting customer churn
* Forecasting sales
* Estimating patient readmission risk
* Identifying fraudulent transactions

Predictive models analyze historical patterns to estimate future probabilities.

### Descriptive Analysis

When the objective is to understand relationships and patterns within data, a descriptive approach is appropriate.

Descriptive analysis helps:

* Identify trends
* Discover behavioral patterns
* Explore relationships between variables
* Understand customer preferences

Clustering techniques are often used to group similar observations based on shared characteristics.

### Statistical Analysis

Statistical methods are useful when questions involve counts, measurements, comparisons, or hypothesis testing.

Examples include:

* Measuring average performance
* Comparing groups
* Identifying significant differences
* Analyzing survey results

Statistical analysis provides quantitative evidence to support decision-making.

### Classification Analysis

When the desired outcome is a yes-or-no answer or category prediction, classification techniques are commonly applied.

Examples include:

* Will a customer buy a product?
* Will a patient be readmitted?
* Is a transaction fraudulent?
* Will a loan applicant default?

Classification models assign observations into predefined categories based on historical data.

### Machine Learning

Machine Learning enables computers to learn patterns from data without being explicitly programmed for every scenario.

Machine learning can:

* Discover hidden relationships
* Detect trends
* Improve predictions over time
* Automate decision-making processes

These capabilities make machine learning valuable for solving complex business problems.

### Clustering and Association Analysis

When the objective is to better understand human behavior or discover natural groupings within data, clustering and association techniques are useful.

These methods help identify:

* Customer segments
* Purchasing patterns
* Behavioral similarities
* Relationships among events and actions

Such insights can support targeted marketing, personalization, and strategic planning.

## Healthcare Case Study: Selecting the Analytic Approach

In the healthcare readmission project, the goal was to predict whether patients with congestive heart failure would be readmitted after treatment.

After evaluating the business requirements, the team selected a **Decision Tree Classification Model**.

This approach was chosen because it met several important requirements:

* Predicting patient outcomes
* Estimating readmission risk
* Explaining prediction results
* Supporting clinical decision-making

## How the Decision Tree Model Works

A decision tree evaluates patient characteristics through a sequence of decision points known as nodes.

Each node examines a specific variable and compares it against a threshold value.

Following the branches of the tree eventually leads to a terminal node, known as a leaf, where the final prediction is made.

The model produces:

* A predicted outcome (Yes or No)
* The probability associated with that outcome

This allows analysts to determine the likelihood of patient readmission for each individual.

## Calculating Readmission Risk

The decision tree estimates risk based on the proportion of outcomes within each leaf node.

### When the Dominant Outcome is Yes

If most patients in a leaf node were readmitted, the readmission risk equals the proportion of "Yes" outcomes within that group.

### When the Dominant Outcome is No

If most patients in a leaf node were not readmitted, the readmission risk is calculated as:

**1 − Proportion of No Outcomes**

This provides a consistent measure of readmission risk across all patient groups.

## Advantages of Decision Trees

The decision tree classification approach offered several benefits for the healthcare project.

### Easy to Understand

Decision trees present results in a visual and intuitive format that healthcare professionals can easily interpret.

### Transparent Decision Process

Clinicians can see exactly which patient characteristics contributed to a high-risk classification.

### Supports New Patient Evaluation

The model can be applied to new patients to estimate their likelihood of readmission.

### Dynamic Risk Assessment

Multiple decision tree models can be used throughout a patient's hospital stay.

This enables clinicians to:

* Monitor changes in patient risk
* Evaluate treatment effectiveness
* Adjust care plans accordingly

The result is a continuously updated view of patient risk as treatment progresses.

## Why the Decision Tree Approach Was Selected

The decision tree classification model was selected because it satisfied all major business requirements.

It provided:

* Accurate prediction of readmission outcomes
* Readmission risk estimation
* Explainable results
* Ease of use for healthcare professionals

Most importantly, it allowed clinicians to understand both the prediction and the reasoning behind it, making it a practical solution for real-world healthcare decision-making.
