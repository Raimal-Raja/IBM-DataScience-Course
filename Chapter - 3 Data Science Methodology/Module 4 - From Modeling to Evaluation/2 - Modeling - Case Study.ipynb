### Modeling – Case Study

#### Introduction

The modeling stage is where data scientists apply analytical techniques to prepared data in order to generate predictions and insights. Once the dataset has been cleaned, transformed, and divided into training and testing sets, different models can be built and refined to achieve the best possible performance.

In this case study, a decision tree classification model was developed to predict the likelihood of congestive heart failure (CHF) patient readmission. A key focus of the modeling process was parameter tuning, which involves adjusting model settings to improve predictive performance.

#### Building the Initial Model

Using the prepared training dataset, the first decision tree classification model was created.

The objective was to identify patients at high risk of readmission, making the target outcome:

**Congestive Heart Failure Readmission = Yes**

The initial model achieved:

* **Overall Accuracy:** 85%
* **Readmission ("Yes") Accuracy:** 45%

Although the overall accuracy appeared strong, the model correctly identified less than half of the actual readmission cases. Since predicting high-risk patients was the primary goal, this performance was considered inadequate.

#### Understanding Classification Errors

To improve the model, the team examined the impact of classification errors.

##### False Positive (Type I Error)

A false positive occurs when a patient who would not be readmitted is incorrectly predicted as high risk.

Consequences include:

* Unnecessary intervention
* Additional healthcare costs
* Wasted resources

##### False Negative (Type II Error)

A false negative occurs when a patient who will be readmitted is incorrectly predicted as low risk.

Consequences include:

* Missed intervention opportunities
* Higher treatment costs
* Increased patient risk and trauma
* Additional hospital readmissions

In healthcare applications, false negatives are often significantly more costly than false positives.

#### Parameter Tuning Through Cost Adjustment

Decision tree algorithms allow the adjustment of misclassification costs.

By default, both error types are treated equally with a:

**1:1 cost ratio**

To place greater importance on identifying actual readmissions, the team increased the cost associated with misclassifying a "Yes" outcome.

#### Second Model: Cost Ratio 9:1

For the second model, the relative misclassification cost was adjusted to:

**9:1**

This heavily penalized missing a readmission case.

The results were:

* **Readmission ("Yes") Accuracy:** 97%
* **Overall Accuracy:** 49%

While the model became extremely effective at identifying readmissions, it generated a very large number of false positives.

As a result:

* Many low-risk patients would receive unnecessary intervention.
* Healthcare resources would be wasted.
* Overall model performance declined significantly.

This demonstrated that maximizing sensitivity alone was not an effective solution.

#### Third Model: Cost Ratio 4:1

To achieve a better balance, the relative cost was adjusted to:

**4:1**

This produced a more balanced outcome:

* **Sensitivity (Yes Accuracy):** 68%
* **Specificity (No Accuracy):** 85%
* **Overall Accuracy:** 81%

These results represented the best compromise between identifying high-risk patients and avoiding excessive false positives.

#### Understanding Sensitivity and Specificity

##### Sensitivity

Sensitivity measures how effectively the model identifies actual positive cases.

In this case:

* Percentage of patients who were truly readmitted and correctly predicted as readmitted.
* Achieved sensitivity: **68%**

##### Specificity

Specificity measures how effectively the model identifies actual negative cases.

In this case:

* Percentage of patients who were not readmitted and correctly predicted as not readmitted.
* Achieved specificity: **85%**

Balancing these two measures is often more important than maximizing overall accuracy alone.

#### Iterative Model Improvement

Model tuning was not limited to adjusting classification costs.

Additional improvements involved:

* Revisiting the data preparation stage
* Refining variable definitions
* Creating improved feature representations
* Testing alternative configurations

This iterative process helped the model better capture the underlying clinical patterns associated with CHF readmissions.

#### Final Outcome

Through parameter tuning and model refinement, the team developed a decision tree model that achieved a practical balance between sensitivity and specificity.

The final model:

* Improved identification of high-risk patients
* Reduced excessive false positives
* Maintained strong overall accuracy
* Provided clinically useful predictions for healthcare professionals

#### Key Takeaways

* Modeling involves building and refining predictive algorithms using prepared data.
* The initial model achieved high overall accuracy but poor readmission detection.
* False positives and false negatives have different costs and consequences.
* Adjusting misclassification costs can significantly influence model behavior.
* A 9:1 cost ratio improved sensitivity but created too many false positives.
* A 4:1 cost ratio produced the best balance between sensitivity and specificity.
* Modeling is an iterative process that often requires returning to earlier stages for further improvements.
* Successful models balance accuracy, sensitivity, and specificity to meet business and operational objectives.
