# Data Science Methodology 101: From Requirements to Collection – Data Requirements

## Introduction

Once the business problem has been clearly defined and an appropriate analytical approach has been selected, the next step in the data science methodology is determining the data requirements. Just as a chef needs the correct ingredients to prepare a successful meal, a data scientist needs the right data to solve a business problem effectively.

The Data Requirements stage focuses on identifying what data is needed, where it can be obtained, how it should be structured, and how it will support the chosen analytical approach.

## Understanding Data Requirements

Data requirements define the information necessary to answer the business question and build the selected analytical model.

Before data can be collected or prepared, data scientists must determine:

* What data is needed
* Where the data will come from
* What format the data should have
* How the data will be represented
* Whether the available data supports the chosen analytical approach

Careful planning at this stage helps avoid problems later in the project lifecycle.

## The Role of Data in the Methodology

The relationship between the business problem and data can be compared to preparing a recipe.

* The business problem is the recipe.
* Data is the collection of ingredients.
* The analytical approach is the cooking method.
* The final model or insight is the completed meal.

Without the proper ingredients, even the best recipe cannot produce the desired outcome.

Similarly, without the correct data, even the most advanced analytical methods will fail to generate meaningful results.

## Defining Data Requirements for Decision Tree Classification

Before beginning data collection and preparation, specific requirements must be established based on the selected analytical technique.

For a decision tree classification model, data scientists must determine:

### Data Content

The information required to describe the problem and support prediction.

### Data Format

The structure in which the data must be organized for modeling.

### Data Sources

The systems, databases, or repositories from which the data will be collected.

Establishing these requirements ensures that the collected data can be effectively used during model development.

## Healthcare Case Study

In the healthcare readmission project, the team selected a decision tree classification approach to predict patient readmission risk.

The first task was to define the data requirements needed to build the model.

This process began by identifying an appropriate patient cohort from the healthcare provider's membership base.

## Patient Cohort Selection Criteria

To create a complete and reliable dataset, the team established three inclusion criteria.

### Inpatient Admission Requirement

Patients needed to have been admitted as inpatients within the healthcare provider's service area.

This ensured that sufficient clinical information would be available for analysis.

### Congestive Heart Failure Diagnosis

Patients were required to have a primary diagnosis of congestive heart failure during a full year of observation.

This allowed the project to focus specifically on the target patient population.

### Continuous Enrollment Requirement

Patients needed to have continuous enrollment for at least six months prior to their primary congestive heart failure admission.

This requirement ensured access to a comprehensive medical history.

## Exclusion Criteria

Not all patients meeting the inclusion requirements were suitable for analysis.

Patients with congestive heart failure who also suffered from other major medical conditions were excluded from the study.

These additional conditions could significantly increase readmission rates and potentially distort the model's findings.

By removing these cases, the team reduced bias and improved the reliability of the analysis.

## Defining Data Content

After selecting the patient cohort, the team identified the information needed to support the decision tree model.

To predict readmission outcomes, data was required from all aspects of each patient's clinical history.

This included:

* Hospital admissions
* Primary diagnoses
* Secondary diagnoses
* Tertiary diagnoses
* Medical procedures
* Prescription records
* Healthcare services received during hospitalization
* Patient-doctor visit records

Collecting this information provided a complete view of each patient's healthcare experience.

## Managing Large Volumes of Data

A single patient could generate thousands of transactional records throughout their medical history.

Examples include:

* Multiple hospital visits
* Numerous prescriptions
* Various procedures
* Ongoing treatments
* Follow-up consultations

While this detailed information was valuable, it needed to be transformed into a format suitable for decision tree classification.

## Data Format Requirements

The selected modeling technique required a specific data structure.

For decision tree classification:

* Each row represented a single patient.
* Each column represented a variable used by the model.

This format differs significantly from raw healthcare transaction data, where multiple records may exist for a single patient.

## Preparing Data for Modeling

To create the required structure, the data scientists aggregated transactional records to the patient level.

This process involved:

* Combining multiple records into a single patient profile
* Creating new variables that summarized patient history
* Transforming raw transactional data into analytical features

The resulting dataset contained one record per patient, making it suitable for decision tree analysis.

## Importance of Planning Ahead

The Data Requirements phase also highlights the importance of anticipating future stages of the methodology.

Understanding how the data will eventually be used allows data scientists to:

* Collect appropriate information
* Design efficient data preparation processes
* Reduce unnecessary work
* Improve model quality

Thinking ahead helps ensure that data collected today can successfully support analysis tomorrow.

## Key Takeaways

* Data requirements define the information needed to solve a business problem.
* Data scientists must identify the necessary data content, format, and sources before collection begins.
* Requirements should align with the selected analytical approach.
* In the healthcare case study, a patient cohort was selected using specific inclusion and exclusion criteria.
* Comprehensive clinical history data was required to predict patient readmission.
* Decision tree classification required one record per patient with variables represented as columns.
* Transactional healthcare records were transformed into patient-level features during data preparation.
* Careful planning during the Data Requirements phase improves efficiency and supports successful model development.
