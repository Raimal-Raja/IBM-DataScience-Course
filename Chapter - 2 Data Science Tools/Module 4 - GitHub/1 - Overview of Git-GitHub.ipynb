## What is Version Control?

Version control is a system that tracks changes made to files over time.

It allows users to:

* Record modifications
* Review previous versions
* Restore older versions
* Collaborate with others
* Track project history

Version control is an essential tool in software development, data science, and many other fields.

---

## Why is Version Control Important?

When multiple people work on the same project, changes can quickly become difficult to manage.

Without version control:

* Files may be overwritten
* Changes may be lost
* Collaboration becomes difficult
* Mistakes are harder to fix

With version control:

* Every change is tracked
* Previous versions are preserved
* Team members can work together efficiently
* Mistakes can be reversed easily

---

## Example of Version Control

Imagine you create a shopping list and ask several roommates to review it.

Without version control:

```text id="git001"
Milk
Bread
Eggs
```

Each person edits the list separately, creating multiple conflicting versions.

You must manually compare and merge everyone's changes.

---

## Version Control Solution

With version control:

```text id="git002"
Shopping List
      ↓
Change Tracking
      ↓
Revision History
      ↓
Final Version
```

Every modification is recorded, making it easy to determine:

* Who made a change
* What was changed
* When it was changed

---

## What is Git?

Git is a free and open-source version control system.

It is distributed under the GNU General Public License (GPL).

Git was designed to:

* Track changes in files
* Support collaboration
* Manage project history
* Enable distributed development

Today, Git is one of the most widely used version control systems in the world.

---

## Distributed Version Control

Git is a **distributed version control system**.

This means every user can have a complete copy of the project on their own computer.

Example:

```text id="git003"
Developer A
      ↓
Local Repository

Developer B
      ↓
Local Repository

Developer C
      ↓
Local Repository
```

Each user works independently and later synchronizes changes.

---

## Local and Remote Repositories

A Git workflow typically includes:

```text id="git004"
Local Repository
       ↓
Remote Repository
```

Developers make changes locally and then share them through a remote repository.

This approach improves reliability because every developer has a complete project history.

---

## Benefits of Distributed Version Control

Advantages include:

* Offline development
* Faster operations
* Redundancy
* Easier collaboration
* Complete project history on every machine

Even if a remote server becomes unavailable, local repositories continue to function.

---

## What Can Git Track?

Git is commonly used for source code, but it can manage many file types.

Examples include:

* Source code
* Documentation
* Images
* Configuration files
* Text files
* Data files

Any file-based project can benefit from version control.

## What is GitHub?

GitHub is a web-based hosting platform for Git repositories.

It provides:

* Repository hosting
* Collaboration tools
* Pull requests
* Issue tracking
* Code reviews
* Project management features

GitHub simplifies teamwork and project sharing.

---

## Other Git Hosting Platforms

GitHub is not the only Git hosting service.

Other popular platforms include:

* GitLab
* Bitbucket
* Beanstalk

All of these platforms provide Git repository hosting with collaboration features.

---

## Using Git Without GitHub

Git can be used entirely from the command line.

Developers can:

* Create repositories
* Track changes
* Create branches
* Merge code

without using any web-based platform.

GitHub simply adds additional collaboration and hosting capabilities.

## Important Git Terminology

Before working with Git, it is important to understand several common terms.

---

## SSH

SSH (Secure Shell) is a protocol used for secure communication between computers.

SSH is commonly used for:

* Remote access
* Authentication
* Secure repository operations

Many developers use SSH keys to connect to Git repositories securely.

---

## Repository

A repository, often called a **repo**, is a project folder that is managed by Git.

A repository contains:

* Project files
* Revision history
* Configuration information

Example:

```text id="git005"
Project Repository
 ├── Code Files
 ├── Documentation
 ├── Images
 └── Git History
```

---

## Fork

A fork is a personal copy of another repository.

Forks allow developers to:

* Experiment safely
* Make changes independently
* Contribute to projects

without affecting the original repository.

---

## Pull Request

A pull request is a request to merge changes into another repository or branch.

Pull requests enable:

* Code review
* Discussion
* Approval workflows
* Collaboration

They are commonly used in open-source development.

---

## Working Directory

The working directory contains the files and folders currently being edited on your computer.

Example:

```text id="git006"
Project Folder
 ├── file1.py
 ├── file2.py
 └── README.md
```

Changes made here are not automatically tracked until Git is instructed to record them.

---

## Git Workflow Overview

A typical Git workflow follows this process:

```text id="git007"
Working Directory
        ↓
Staging Area
        ↓
Repository
```

Changes move through each stage before becoming part of the project's history.

---

## Creating a Repository

To begin using Git, a repository must first be created.

Git provides:

```bash id="git008"
git init
```

This command initializes a new repository.

It creates the necessary Git metadata and enables version control.

---

## Cloning a Repository

Another common way to start is by copying an existing repository.

This is called cloning.

Example:

```bash id="git009"
git clone <repository-url>
```

Cloning downloads the repository and its complete history.

---

## Adding Changes

After modifying files, Git must be informed about the changes.

Example:

```bash id="git010"
git add filename
```

The `git add` command moves changes into the staging area.

---

## The Staging Area

The staging area acts as an intermediate step between:

```text id="git011"
Working Directory
        ↓
Staging Area
        ↓
Repository
```

It allows developers to choose exactly which changes will be included in a commit.

---

## Checking Repository Status

To view the current state of the repository:

```bash id="git012"
git status
```

This command shows:

* Modified files
* Staged files
* Untracked files
* Current branch information

It is one of the most frequently used Git commands.

---

## Creating a Commit

A commit records a snapshot of the project.

Example:

```bash id="git013"
git commit -m "Added new feature"
```

A commit permanently stores the staged changes in the repository history.

---

## Why Commits Matter

Commits provide:

* Change history
* Project milestones
* Accountability
* Easy rollback capabilities

Each commit represents a specific point in the project's development.

---

## Undoing Changes

Sometimes changes need to be discarded.

Git provides:

```bash id="git014"
git reset
```

This command can be used to undo modifications or adjust staged changes.

---

## Viewing History

To review previous commits:

```bash id="git015"
git log
```

The log displays:

* Commit identifiers
* Authors
* Dates
* Commit messages

This provides a complete history of project development.

---

## Branches

Branches allow developers to work on changes without affecting the main project.

Example:

```bash id="git016"
git branch feature1
```

This creates a separate development path.

---

## Why Use Branches?

Branches are useful for:

* New features
* Bug fixes
* Experiments
* Team collaboration

They isolate changes until they are ready to be merged.

---

## Switching Between Branches

To move between branches:

```bash id="git017"
git checkout branch-name
```

This command changes the active branch.

Developers can work on multiple features independently.

---

## Merging Branches

After development is complete, branches can be merged.

Example:

```bash id="git018"
git merge feature1
```

This combines the branch changes into the current branch.

---

## Branch Workflow

A typical branch workflow looks like:

```text id="git019"
Main Branch
      │
      ├── Feature Branch
      │
      └── Bug Fix Branch
```

After development:

```text id="git020"
Feature Branch
       ↓
Merge
       ↓
Main Branch
```

---

## Git Collaboration Workflow

Git enables collaboration among multiple developers.

Example:

```text id="git021"
Developer A
Developer B
Developer C
       ↓
Shared Repository
```

Each developer works independently and shares changes through Git.

---

## Common Git Commands

### Initialize Repository

```bash id="git022"
git init
```
