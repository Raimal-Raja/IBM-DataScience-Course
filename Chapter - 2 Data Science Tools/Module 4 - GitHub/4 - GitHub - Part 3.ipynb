---

## Contributing to Existing Repositories

When working with open-source projects, you may want to contribute improvements or bug fixes to repositories maintained by other people or organizations.

For example:

* Fixing a typo
* Improving documentation
* Adding a feature
* Fixing a bug
* Improving performance

In many cases, contributors do not have direct write access to the original repository.

To solve this problem, GitHub provides a feature called **Forking**.

---

## What is a Fork?

A **fork** is a personal copy of another user's repository.

Forking allows you to:

* Create your own copy of a project
* Experiment safely
* Make changes independently
* Submit improvements back to the original project

The original repository remains unchanged until your changes are accepted.

---

## Why Use Forks?

Instead of requesting direct commit access to a repository, contributors can:

1. Fork the repository.
2. Make changes in their own copy.
3. Submit a pull request.
4. Wait for the maintainers to review the changes.

This workflow is commonly used in open-source development.

---

## Forking a Repository

Suppose you want to contribute to a repository hosted by an organization.

The repository might look like:

```text id="fork001"
Organization/Project
```

To create your own copy:

1. Open the repository on GitHub.
2. Click the **Fork** button.
3. Select your GitHub account.

GitHub creates a copy of the repository under your account.

Example:

```text id="fork002"
Original Repository:
IBM/Coursera

Forked Repository:
username/Coursera
```

GitHub also indicates the source repository.

Example:

```text id="fork003"
forked from IBM/Coursera
```

---

## Working with a Fork

After creating a fork, you can make changes directly on GitHub.

However, many developers prefer working locally.

The first step is cloning the forked repository.

---

## Cloning a Fork

Clone the repository using Git:

```bash id="fork004"
git clone git@github.com:username/Coursera.git
```

This creates a local copy of the forked repository.

Repository structure:

```text id="fork005"
Remote Repository (Fork)
        ↓
Local Repository
```

You can now edit files locally.

---

## SSH Authentication

Many developers use SSH to authenticate with GitHub.

SSH uses a pair of keys:

### Private Key

```text id="fork006"
id_rsa
```

* Stored securely on your computer.
* Must never be shared.

### Public Key

```text id="fork007"
id_rsa.pub
```

* Safe to share.
* Added to GitHub.

---

## Adding an SSH Key to GitHub

To use SSH authentication:

1. Generate an SSH key pair.
2. Copy the contents of the public key.
3. Open GitHub Settings.
4. Navigate to:

```text id="fork008"
Settings
  └── SSH and GPG Keys
```

5. Click:

```text id="fork009"
New SSH Key
```

6. Paste the public key.
7. Save the key.

GitHub can now verify your identity using your private key.

---

## Making Changes to a Fork

Suppose we want to improve a README file.

Original text:

```text id="fork010"
coursera specialization
```

Corrected version:

```text id="fork011"
Coursera specialization
```

A small typo has been fixed.

---

## Staging Changes

After editing the file, stage the changes:

```bash id="fork012"
git add README.md
```

Check the status:

```bash id="fork013"
git status
```

Git reports that the README file has been modified.

---

## Creating a Commit

Create a meaningful commit message:

```bash id="fork014"
git commit -m "Corrected a typo in README"
```

Good commit messages should clearly describe the purpose of the change.

Examples:

```text id="fork015"
Corrected typo in README
Updated installation instructions
Fixed broken link
Added documentation
```

---

## Pushing Changes

Push the commit to your fork:

```bash id="fork016"
git push
```

The changes are uploaded to your GitHub repository.

At this point:

```text id="fork017"
Original Repository  → Unchanged
Forked Repository    → Updated
```

Only your fork contains the new changes.

---

## Viewing Changes on GitHub

After pushing:

* The modified file appears in your repository.
* The commit history is updated.
* GitHub displays the commit message.

Example:

```text id="fork018"
Corrected a typo in README
```

This makes it easy to track contributions.

---

## What is a Pull Request?

A **Pull Request (PR)** is a request to merge changes from one repository or branch into another.

A pull request allows maintainers to:

* Review code
* Discuss changes
* Request improvements
* Accept or reject contributions

Pull requests are the primary method for contributing to open-source projects.

---

## Creating a Pull Request

After pushing changes to your fork:

1. Open your fork on GitHub.
2. Click **Pull Request**.
3. Compare your changes with the original repository.

GitHub automatically identifies differences between:

```text id="fork019"
Fork Repository
       ↓
Original Repository
```

In this example, the only difference is a corrected typo.

---

## Submitting the Pull Request

Create the pull request by providing:

### Title

```text id="fork020"
Corrected typo in README
```

### Description

Explain:

* What changed
* Why the change was made
* Any relevant details

Submit the pull request.

GitHub notifies the maintainers of the original repository.

---

## Reviewing a Pull Request

Repository maintainers can review:

* Modified files
* Commit history
* Proposed changes

For example:

```text id="fork021"
Original:
coursera specialization

Updated:
Coursera specialization
```

The maintainer determines whether the contribution should be accepted.

---

## Merging a Pull Request

If the changes are acceptable, the maintainer can merge the pull request.

Typical workflow:

```text id="fork022"
Contributor
     ↓
Pull Request
     ↓
Review
     ↓
Merge
```

GitHub provides a **Merge Pull Request** button that combines the contribution with the original repository.

---

## Confirming the Merge

After reviewing:

```text id="fork023"
Merge Pull Request
```

The maintainer confirms the merge.

The changes are incorporated into the official repository.

Example:

```text id="fork024"
README.md
```

now contains:

```text id="fork025"
Coursera Specialization
```

The improvement becomes part of the project history.

---

## Contribution Workflow Overview

The complete contribution process looks like this:

```text id="fork026"
Original Repository
        │
        ├── Fork
        │
Forked Repository
        │
        ├── Clone
        ├── Modify Files
        ├── Commit Changes
        └── Push Changes
        │
        └── Pull Request
                │
                ├── Review
                └── Merge
                        │
                        ▼
               Original Repository
```
