# Working with Remote Changes and Branches in Git

In this notebook, we will discuss:

* Synchronizing changes from a remote repository
* Pulling updates from GitHub
* Understanding branches
* Creating and switching branches
* Making changes in a branch
* Pushing branches to GitHub
* Merging branches into the main branch

---

## Synchronizing with a Remote Repository

When working with Git, other developers may make changes to the same repository.

These changes are stored in the remote repository but are not automatically available in your local repository.

To stay up to date, you need to synchronize your local repository with the remote repository.

---

## Creating and Pushing Initial Content

Suppose we create a README file:

```text id="git101"
README.md
```

Stage the file:

```bash id="git102"
git add README.md
```

Create a commit:

```bash id="git103"
git commit -m "Initial commit"
```

Push the changes to GitHub:

```bash id="git104"
git push
```

After the push is complete, the README file becomes available in the remote repository.

---

## Remote Changes

Now imagine that someone else adds a file directly to the remote repository.

For example:

```text id="git105"
newfile.txt
```

At this point:

### Remote Repository

```text id="git106"
README.md
newfile.txt
```

### Local Repository

```text id="git107"
README.md
```

The new file exists only in the remote repository.

---

## Pulling Changes from GitHub

To download the latest changes from the remote repository, use:

```bash id="git108"
git pull
```

Git retrieves the latest updates and merges them into your local repository.

After running the command:

```text id="git109"
README.md
newfile.txt
```

Your local repository now matches the remote repository.

---

## Understanding Branches

Branches allow developers to work on new features or experiments without affecting the main project.

The default branch is commonly called:

```text id="git110"
master
```

or in newer repositories:

```text id="git111"
main
```

A branch creates an independent line of development.

---

## Why Use Branches?

Branches help developers:

* Work on new features safely
* Experiment without affecting production code
* Collaborate with multiple team members
* Keep the main branch stable

Each branch maintains its own set of changes until it is merged.

---

## Creating a New Branch

To create a new branch:

```bash id="git112"
git branch mybranch
```

This creates a branch named:

```text id="git113"
mybranch
```

However, creating a branch does not automatically switch to it.

---

## Checking the Current Branch

To see your current branch:

```bash id="git114"
git status
```

Example output:

```text id="git115"
On branch master
```

This indicates that you are still working on the master branch.

---

## Switching to a Branch

To switch to the new branch:

```bash id="git116"
git checkout mybranch
```

Verify the change:

```bash id="git117"
git status
```

Example output:

```text id="git118"
On branch mybranch
```

You are now working inside the new branch.

---

## Making Changes in a Branch

Suppose we create a file:

```text id="git119"
stuffonbranch.txt
```

with the content:

```text id="git120"
Stuff on branch
```

Stage the file:

```bash id="git121"
git add stuffonbranch.txt
```

Create a commit:

```bash id="git122"
git commit -m "Added branch file"
```

The changes now exist only in the branch.

---

## Pushing a Branch to GitHub

Before others can access the branch, it must be pushed to GitHub.

Push the branch:

```bash id="git123"
git push -u origin mybranch
```

This creates a corresponding remote branch.

After the push:

```text id="git124"
Remote Repository
├── master
└── mybranch
```

The file exists in `mybranch` but not yet in `master`.

---

## Comparing Branches

### Master Branch

```text id="git125"
README.md
newfile.txt
```

### MyBranch

```text id="git126"
README.md
newfile.txt
stuffonbranch.txt
```

The additional file exists only in the branch.

---

## Returning to the Main Branch

To switch back to the main branch:

```bash id="git127"
git checkout master
```

Check the current branch:

```bash id="git128"
git status
```

Output:

```text id="git129"
On branch master
```

You are now back on the master branch.

---

## Merging a Branch

To merge changes from the branch into the master branch:

```bash id="git130"
git merge mybranch
```

Git combines the branch changes with the master branch.

After the merge, the new file becomes available in your local master branch.

---

## Local Merge vs Remote Merge

After merging:

### Local Master Branch

```text id="git131"
README.md
newfile.txt
stuffonbranch.txt
```

### Remote Master Branch

```text id="git132"
README.md
newfile.txt
```

The merge exists only locally at this stage.

---

## Pushing the Merge to GitHub

To update the remote master branch:

```bash id="git133"
git push
```

Git uploads the merged changes to GitHub.

Now both repositories contain:

```text id="git134"
README.md
newfile.txt
stuffonbranch.txt
```

The branch changes have successfully become part of the main project.

---

## Branch Workflow Overview

A typical branching workflow looks like this:

```text id="git135"
master
   │
   ├── create branch
   │
mybranch
   │
   ├── make changes
   ├── commit changes
   └── push branch
   │
   └── merge into master
           │
           └── push master
```

This workflow enables developers to work independently while keeping the main branch stable.

---

## Common Commands for Branch Management

### Create a Branch

```bash id="git136"
git branch mybranch
```

```bash id="git137"

```

### Check Current Branch

```bash id="git138"
git status
```


### Push Changes

```bash id="git141"

```

### Push a New Branch

```bash id="git142"
git push -u origin mybranch
```
