## What is Git?

**Git** is a distributed version control system used to track changes in files and collaborate with others on software projects.

Git allows developers to:

* Track changes over time
* Collaborate with team members
* Maintain different versions of a project
* Restore previous versions when needed

Most software development projects use Git for version control.

---

## What is GitHub?

GitHub is a cloud-based platform that hosts Git repositories.

It provides:

* Remote storage for repositories
* Collaboration tools
* Project management features
* Code review capabilities
* Version history tracking

GitHub allows multiple developers to work together on the same project.

---

## GitHub Alternatives

GitHub is not the only platform that supports Git repositories.

One popular alternative is **GitLab**.

GitLab offers:

* Git repository hosting
* Collaboration features
* Open-source deployment options
* Self-hosted installations

Organizations can run GitLab on their own infrastructure while still benefiting from version control and collaboration tools.

---

## Understanding Repositories

A **repository** (often called a repo) is a storage location for your project files and version history.

A repository contains:

* Project files
* Commit history
* Branches
* Configuration information

Repositories can be either:

### Public Repositories

Anyone can view the repository.

### Private Repositories

Access is restricted to authorized users.

---

## Creating a GitHub Repository

When creating a repository on GitHub, you can choose options such as:

* Repository name
* Public or private visibility
* Initialize with a README file
* Add a `.gitignore` file

For example:

```text id="g1t001"
Repository Name: demo
Visibility: Public
Initialize with README: Yes
```

Once the repository is created, GitHub hosts it remotely.

---

## Cloning a Repository

To work with a repository on your local machine, you create a copy using the **git clone** command.

### Clone Using SSH

```bash id="git001"
git clone git@github.com:username/demo.git
```

### Clone Using HTTPS

```bash id="git002"
git clone https://github.com/username/demo.git
```

Cloning copies the repository contents from GitHub to your local computer.

---

## SSH Authentication

Many developers prefer using SSH because it simplifies authentication.

To use SSH:

1. Generate an SSH key pair.
2. Add the public key to GitHub.
3. Use the SSH repository URL when cloning or pushing changes.

This eliminates the need to repeatedly enter GitHub credentials.

---

## Repository Structure

After cloning a repository, you may see files such as:

```text id="git003"
README.md
.git/
```

### README.md

Contains project documentation.

### .git Directory

Stores Git's internal repository information.

**Important:** Never manually modify files inside the `.git` directory.

---

## Creating New Files

Suppose we create a file called:

```text id="git004"
test.txt
```

and add the following content:

```text id="git005"
Hello, this is a test.
```

At this point, Git knows the file exists but is not yet tracking it.

---

## The Staging Area

Before saving changes to the repository, files must be added to the staging area.

### Add a File

```bash id="git006"
git add test.txt
```

To check the repository status:

```bash id="git007"
git status
```

Git will show that the file is staged and ready to be committed.

---

## Understanding the Git Workflow

Git uses three primary locations:

### Working Directory

The files currently on your computer.

### Staging Area

A temporary area where changes are prepared for a commit.

### Local Repository

The version-controlled history stored on your machine.

A simplified workflow looks like:

```text id="git008"
Working Directory
       ↓
 Staging Area
       ↓
 Local Repository
       ↓
 Remote Repository
```

---

## Adding Multiple Files

Suppose we create another file:

```text id="git009"
test2.txt
```

We can stage both files:

```bash id="git010"
git add test.txt
git add test2.txt
```

Running:

```bash id="git011"
git status
```

shows both files in the staging area.

---

## Creating a Commit

A **commit** is a snapshot of your project at a specific point in time.

To create a commit:

```bash id="git012"
git commit -m "Initial commit"
```

The commit is stored in your local repository.

---

## Why Commits Matter

Commits are the core building blocks of Git.

Each commit:

* Records changes
* Stores project history
* Allows rollback to previous versions
* Facilitates collaboration

Developers often group related changes into a single commit.

This makes project history easier to understand.

---

## Local vs Remote Repositories

After committing, the changes exist only on your local machine.

GitHub still does not know about these changes.

```text id="git013"
Local Repository ✓
Remote Repository ✗
```

To share changes, you must push them to GitHub.

---

## Pushing Changes to GitHub

To upload commits to the remote repository:

```bash id="git014"
git push
```

Git transfers commits from the local repository to GitHub.

After pushing, the new files become visible in the GitHub repository.

---

## Viewing Changes on GitHub

Once changes are pushed:

* Files appear in the repository.
* Commit history is updated.
* GitHub displays which commit added or modified each file.

This makes it easy to track project evolution over time.

---

## Initializing a Repository Locally

Instead of creating a repository on GitHub first, you can start locally.

Create a project folder:

```bash id="git015"
mkdir demo
cd demo
```

Initialize Git:

```bash id="git016"
git init
```

This creates the `.git` directory and turns the folder into a Git repository.

---

## Adding Files to a New Repository

Create a README file:

```text id="git017"
README.md
```

Add all files:

```bash id="git018"
git add .
```

The dot (`.`) means "add everything in the current directory."

---

## Creating the First Commit

Commit the files:

```bash id="git019"
git commit -m "Initial commit"
```

The project now contains a local Git history.

---

## Connecting to GitHub

At this point, Git still does not know where to push the repository.

A remote repository must be configured.

### Add a Remote Repository

```bash id="git020"
git remote add origin git@github.com:username/demo.git
```

Here:

* `origin` is the remote name.
* The URL points to the GitHub repository.

---

## Pushing for the First Time

After adding the remote, push the repository:

```bash id="git021"
git push -u origin master
```

The `-u` flag establishes a tracking relationship between the local branch and the remote branch.

Future pushes can be performed with:

```bash id="git022"
git push
```

---

## Common Git Commands

### Clone a Repository

```bash id="git023"
git clone <repository-url>
```
