## What is R?

R is an open-source programming language designed primarily for:

* Statistical analysis
* Data visualization
* Data science
* Machine learning
* Research computing

R originated from the S programming language and has become one of the most widely used tools for statistical computing.

---

## What is RStudio?

RStudio is an Integrated Development Environment (IDE) designed specifically for R.

It provides a user-friendly interface for:

* Writing code
* Running scripts
* Managing data
* Creating visualizations
* Developing statistical models

RStudio helps users work more efficiently by combining multiple tools within a single application.

---

## Open Source Software

Both R and RStudio are open-source software.

Benefits include:

* Free to use
* Large community support
* Extensive package ecosystem
* Continuous development and improvement

This has contributed to their popularity in academia, research, and industry.

## The Data Frame

The most commonly used data structure in R is the **data frame**.

A data frame stores information in a tabular format consisting of:

* Rows
* Columns

Similar to a spreadsheet or database table.

Example:

```text id="rs003"
Name       Age
Alice      25
Bob        30
Charlie    28
```

Each row represents an observation, while each column represents a variable.

---

## Checking the Working Directory

Before loading data, it is often useful to determine the current working directory.

R provides the following function:

```r id="rs004"
getwd()
```

The function returns the location where R is currently operating.

Example output:

```text id="rs005"
/home/user
```

---

## Why Working Directories Matter

The working directory determines where R:

* Searches for files
* Saves output
* Creates new files

Understanding the working directory helps avoid file path errors.

---

## Reading Data from CSV Files

One of the most common ways to load data into R is through CSV files.

R provides the `read.csv()` function.

Example:

```r id="rs006"
data_frame <- read.csv("customer_messages.csv")
```

This command loads the contents of the CSV file into a data frame.

R supports assignment using:

The `<-` operator is commonly used in R programming.

## Understanding Headers

By default, `read.csv()` assumes that the first row contains column names.

Example CSV:

```text id="rs011"
Name,Age
Alice,25
Bob,30
```

R interprets:

```text id="rs012"
Name
Age
```

as column headers.

---

## When Data Does Not Contain Headers

Sometimes the first row contains actual data rather than column names.

In such cases:

```r id="rs013"
read.csv("customer_messages.csv", header = FALSE)
```

The `header = FALSE` parameter instructs R to treat the first row as data.

---

## Comparing Header Options

### Default Behavior

```r id="rs014"
read.csv("file.csv")
```

First row becomes column names.

### Header Disabled

```r id="rs015"
read.csv("file.csv", header = FALSE)
```

First row becomes part of the dataset.

---

## Function Assistance and Auto-Completion

RStudio provides intelligent code completion.

Benefits include:

* Function suggestions
* Parameter hints
* Syntax assistance

For example, when typing:

```r id="rs016"
read.csv(
```
