## What is an API?

An **API (Application Programming Interface)** allows two pieces of software to communicate with each other.

For example, imagine you have:

* Your program
* Some data
* Other software components

An API acts as a bridge that enables your program to interact with those software components.

One important aspect of APIs is that you do **not** need to understand how the API works internally. You only need to know:

* The inputs it accepts
* The outputs it returns

The API refers only to the interface that users interact with, while the library refers to the complete collection of software components behind the scenes.

---

## API Libraries

Consider the **pandas** library.

Pandas is actually a collection of software components, many of which are not written in Python. When working with data, we use the pandas API to communicate with these components and perform data processing tasks.

### Example Workflow

Data → Pandas API → Software Components → Processed Data

The API provides a simple interface while hiding the complexity of the underlying implementation.

---

## APIs in Different Programming Languages

Sometimes a software system has one backend implementation but multiple APIs.

Consider **TensorFlow**, which is primarily written in C++.

TensorFlow provides APIs for several programming languages, including:

* Python
* JavaScript
* C++
* Java
* Go

In addition, many volunteer-developed APIs exist for languages such as:

* Julia
* MATLAB
* R
* Scala

Regardless of the programming language, the API remains the interface through which developers interact with TensorFlow.

## What are REST APIs?

**REST (Representational State Transfer) APIs** are a popular type of API that enable communication over the internet.

REST APIs provide access to:

* Cloud storage
* Large datasets
* Artificial Intelligence services
* Web applications
* Many other online resources

In a REST API system:

* Your application is called the **Client**
* The online service is called the **Resource**
* Communication occurs through the internet

REST APIs follow a set of rules that define how requests and responses are exchanged.

---

## Common REST API Terminology

### Client

The user or application making the request.

### Resource

The web service or data being accessed.

### Endpoint

The URL used to access the resource.

### Request

A message sent by the client asking the service to perform an action.

### Response

The result returned by the service after processing the request.

---

## HTTP Methods and Communication

REST APIs communicate using **HTTP (Hypertext Transfer Protocol)**.

When interacting with a REST API:

1. The client sends an HTTP request.
2. The request often contains data in **JSON (JavaScript Object Notation)** format.
3. The request is transmitted over the internet.
4. The web service processes the request.
5. The service returns an HTTP response.
6. The response often contains JSON data.

### Request Flow

Client → HTTP Request → Web Service

### Response Flow

Web Service → HTTP Response → Client

---

## JSON in REST APIs

JSON is a lightweight format used for exchanging information between systems.

### Example JSON Request

```json
{
  "text": "Hello World"
}
```

### Example JSON Response

```json
{
  "translation": "Hola Mundo"
}
```

JSON makes it easy for applications to exchange structured data.

## Example 1: Watson Speech-to-Text API

The Watson Speech-to-Text API converts spoken audio into text.

### Process

1. The client sends an audio file to the API.
2. This is typically done using a **POST request**.
3. The API processes the audio.
4. The API returns the transcribed text.

### Workflow

Audio File → API Request → Speech-to-Text Service → Text Response

---

## Example 2: Watson Language Translator API

The Watson Language Translator API translates text between languages.

### Process

1. The client sends text to the API.
2. The API receives the request.
3. The text is translated.
4. The translated text is returned to the client.

### Example

English:

```text
Hello, how are you?
```

Spanish:

```text
Hola, ¿cómo estás?
```

### Workflow

English Text → Translation API → Spanish Text
