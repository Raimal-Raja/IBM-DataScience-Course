## Why Was DAX Created?

Although many open datasets are available today, finding datasets that are both:

* High quality
* Properly licensed
* Suitable for enterprise use

can still be challenging.

To address this issue, IBM created the **Data Asset eXchange (DAX)**.

DAX serves as a trusted repository that helps users discover and access curated open datasets with clearly defined licensing and usage terms.

---

## What is DAX?

The **Data Asset eXchange (DAX)** is an open data repository developed by IBM.

It provides access to datasets that are ready for use in:

* Research projects
* Data science applications
* Machine learning workflows
* Enterprise solutions

DAX focuses on providing datasets that are easy to discover, understand, and use.

## Curated Data Quality

One of the major advantages of DAX is its emphasis on dataset quality.

Datasets are curated to help ensure:

* Better documentation

## Licensing and Usage Terms

Understanding dataset licensing is essential before using data in projects.

DAX helps simplify this process by providing:

* Clearly defined licensing information
* Usage guidelines
* Dataset metadata

Whenever possible, DAX makes datasets available under variants of the **Community Data License Agreement (CDLA)**.

This approach encourages:

* Collaboration
* Data sharing
* Open innovation

while maintaining clear legal guidelines for dataset usage.

As a result, DAX serves as a valuable resource for researchers and developers seeking specialized datasets.

## Exploring a Dataset in DAX

When you find a dataset of interest on DAX, several resources are available.

### Review Metadata

* Dataset format
* File size
* Licensing terms
* Dataset description

Users can preview portions of the dataset before downloading.

This helps determine whether the dataset meets project requirements.
