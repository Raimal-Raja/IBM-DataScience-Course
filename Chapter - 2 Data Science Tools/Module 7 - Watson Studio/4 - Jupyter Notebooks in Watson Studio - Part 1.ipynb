## Working with Jupyter Notebooks in Watson Studio

Jupyter Notebooks are one of the most widely used tools within Watson Studio for data exploration, analysis, visualization, and machine learning development. They provide an interactive environment where code, documentation, visualizations, and results can be combined into a single workspace.

### Creating a Notebook

To begin working with a notebook in Watson Studio:

1. Open an existing project.
2. Select **Add to Project**.
3. Choose **Notebook**.
4. Provide:

   * A Notebook Name
   * An Optional Description
5. Create the notebook.

The notebook is then added to the project and becomes available for development and analysis.

### Adding Data to a Notebook

Before performing analysis, a dataset must be loaded into the project.

The process involves:

1. Opening the **Files** panel.
2. Browsing and selecting the desired file.
3. Uploading the file to the project.

Once uploaded, the dataset becomes available for use within the notebook.

Watson Studio simplifies data access by allowing users to:

* Insert code automatically
* Generate Pandas DataFrame loading commands
* Connect directly to cloud storage resources

This eliminates much of the manual coding normally required to load datasets.

### Documenting the Notebook

Before executing any code, it is considered best practice to document the notebook’s purpose.

To do this:

1. Insert a new cell at the top of the notebook.
2. Change the cell type from **Code** to **Markdown**.
3. Add a description explaining:

   * The notebook objective
   * The dataset being used
   * The analysis being performed
   * Expected outputs

Markdown cells improve readability and make notebooks easier to understand and maintain.

### Running the Notebook

After the dataset and documentation are prepared, the notebook can be executed.

The automatically generated code typically:

* Connects to Cloud Object Storage
* Authenticates using project credentials
* Loads the dataset into a Pandas DataFrame
* Displays the first few records of the dataset

A common first step is reviewing the first five rows to verify that the data has loaded correctly.

### Saving Notebook Changes

After running and modifying the notebook:

* Save the notebook regularly.
* Ensure all changes are committed before leaving the environment.

Saving preserves code, outputs, visualizations, and documentation for future use.

### Accessing Notebooks from Project Assets

All notebooks are stored within the project's **Assets** section.

From the Assets tab, users can:

* Open notebooks
* Review existing work
* Manage notebook resources

When opened from the project, notebooks may initially appear in read-only mode, but editing can be enabled when modifications are required.

### Editing Notebook Properties

Watson Studio provides an information panel that allows users to modify notebook settings.

Available options include:

#### Notebook Information

* Rename the notebook
* Update descriptions
* Review notebook metadata

#### Environment Configuration

Users can:

* Change the runtime environment
* Select different compute resources
* Stop active runtimes
* Restart notebook environments

These capabilities help optimize performance for different workloads.

### Sharing Notebooks

Watson Studio allows notebooks to be shared with others in a read-only format.

Sharing options include:

* Public or restricted links
* Social media sharing
* Selective content sharing

Users can choose how much of the notebook content is visible before generating a shareable version.

This feature is useful for:

* Presentations
* Collaboration
* Knowledge sharing
* Publishing analytical results

### Scheduling Notebook Execution

A powerful Watson Studio feature is the ability to schedule notebooks to run automatically.

To create a scheduled execution:

1. Create a new Job.
2. Assign a job name.
3. Configure scheduling options.

Scheduling settings may include:

* Specific execution dates
* Recurring schedules
* Daily runs
* Weekly runs
* Monthly runs

This functionality is particularly useful for:

* Automated reporting
* Data refresh processes
* Model retraining
* Scheduled analytics workflows

### Monitoring Notebook Jobs

Once a notebook job has been created and executed, users can monitor its progress from the project's **Jobs** tab.

The Jobs section provides visibility into:

* Job Status
* Execution History
* Success or Failure Results
* Scheduled Runs
* Runtime Information

This helps users manage automated workflows and ensure notebooks are executing as expected.

### Key Benefits of Jupyter Notebooks in Watson Studio

Jupyter Notebooks within Watson Studio provide:

* Interactive data analysis
* Integrated documentation using Markdown
* Automated dataset loading
* Pandas DataFrame integration
* Cloud storage connectivity
* Custom runtime environments
* Notebook sharing capabilities
* Automated scheduling and execution
* Centralized project management

By combining code, documentation, data access, and automation within a single environment, Watson Studio enables data scientists and analysts to develop reproducible, collaborative, and scalable data science workflows efficiently.
