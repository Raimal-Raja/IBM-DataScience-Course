# Watson Studio GitHub Integration

## Overview

IBM Watson Studio provides built-in integration with GitHub, enabling users to publish notebooks directly to repositories and collaborate using industry-standard version control workflows. By connecting a GitHub account to Watson Studio, data scientists can manage notebook versions, share projects, track changes, and collaborate with team members more effectively.

This integration supports publishing notebooks to repositories as well as sharing work through GitHub Gists.

---

## Connecting Watson Studio to GitHub

Before publishing notebooks, GitHub integration must be configured.

### Access Profile Settings

To begin:

* Navigate to Watson Studio profile settings.
* Open the **Integrations** tab.
* Access the GitHub integration configuration page.

### Generate a Personal Access Token

GitHub authentication requires a Personal Access Token (PAT).

Steps include:

1. Open the GitHub token generation page.
2. Create a new token.
3. Provide a descriptive token name.
4. Select the **repo** scope permissions.
5. Generate the token.
6. Copy the generated token.

### Configure GitHub Integration

After generating the token:

* Return to Watson Studio.
* Paste the token into the GitHub integration settings.
* Save the configuration.

Watson Studio validates the token automatically before enabling the integration.

---

## Enabling GitHub Integration for a Project

GitHub integration is configured at the project level.

### Project Configuration Steps

1. Open a Watson Studio project.
2. Navigate to the **Settings** tab.
3. Scroll to the GitHub Integration section.
4. Enter the GitHub repository URL.
5. Validate the repository address.
6. Click **Connect**.

Once connected, project assets can be published directly to the selected repository.

## Preparing a Notebook for Publication

Before publishing notebooks to GitHub, sensitive information should be removed.

### Security Best Practices

Ensure that:

* Credentials are removed.
* API keys are hidden.
* Authentication information is replaced with placeholders.
* Sensitive project data is excluded.

A common practice is replacing credentials with masked values before publication.

### Benefits

* Improved security.
* Protection of cloud resources.
* Safe public sharing.
* Reduced risk of credential exposure.

## Publishing Notebooks to GitHub

Watson Studio allows notebooks to be published directly from the notebook interface.

### Publishing Process

1. Open the notebook.
2. Select the publish option.
3. Specify the target repository path.
4. Enter a commit message.
5. Configure publishing options.
6. Publish the notebook.

### Optional Publishing Settings

Users may choose to:

* Include all notebook content.
* Exclude hidden code cells.
* Publish only visible content.

Hidden cells are identified by special notebook comments and are omitted when this option is selected.

## Publishing as a GitHub Gist

Watson Studio also supports publishing notebooks as GitHub Gists.

### What Is a Gist?

A GitHub Gist is a lightweight repository used for sharing code snippets, notebooks, and other files.

Each Gist:

* Is backed by Git.
* Can be cloned.
* Can be forked.
* Maintains revision history.

### Types of Gists

#### Public Gist

* Visible to everyone.
* Searchable by others.
* Suitable for open sharing.

#### Secret Gist

* Accessible only through the URL.
* Not publicly listed.
* Useful for limited distribution.

A secret Gist can later be converted into a public Gist if desired.
