Jupyter Notebooks are one of the most widely used tools for data science, enabling users to combine code, data, visualizations, and documentation within a single interactive workspace. IBM Watson Studio simplifies notebook creation by providing integrated environments, data asset management, and configurable runtime options.

This lesson demonstrates how to create a Jupyter Notebook, upload and access data assets, configure runtime environments, and customize notebook execution settings within Watson Studio.

---

## Uploading Data Assets

Before creating a notebook, data must be available within the project.

### Data Upload Process

* Browse and select files from your local system.
* Alternatively, drag and drop files directly into the project.
* Uploaded files are automatically stored in cloud object storage.
* Once uploaded, the files become project data assets accessible from notebooks and other tools.

### Benefits

* Centralized data storage.
* Easy access across project resources.
* Secure cloud-based management.
* Immediate integration with notebooks.

---

## Creating a Jupyter Notebook

After uploading data, a notebook can be created within the project.

### Notebook Configuration

When creating a notebook, users provide:

* Notebook name
* Description
* Runtime environment
* Programming language
* Processing engine version

### Default Environment

For beginners, Watson Studio provides a default Spark Python environment that includes:

* Python support
* Apache Spark integration
* Preconfigured libraries
* Managed runtime resources

After configuration, Watson Studio automatically provisions the required runtime environment.

---

## Runtime Environment Initialization

Once the notebook is created:

* Watson Studio launches the selected runtime.
* Resources are allocated automatically.
* Required software packages are loaded.
* The notebook becomes ready for development and execution.

This process eliminates manual setup and infrastructure management.

---

## Working with Data Inside a Notebook

After the runtime is active, project data assets can be accessed directly.

### Data Insertion Options

Users can:

* Open the Data Sources panel.
* Locate uploaded files.
* Select **Insert to Code**.
* Choose an appropriate data-loading method.

Available insertion options depend on:

* Notebook language
* File format
* Runtime environment

### Automatic Credential Handling

Generated code automatically includes:

* Object storage connection details.
* Authentication credentials.
* File access configuration.

This simplifies data loading and removes the need for manual connection setup.

---

## Data Preview and Exploration

When the generated code is executed:

* The dataset is loaded into a DataFrame.
* The first five rows are displayed automatically.
* Users can immediately inspect the structure and contents of the dataset.

### Advantages

* Quick validation of data imports.
* Faster exploratory analysis.
* Reduced coding effort.

## Creating Custom Environments

Custom environments provide flexibility for specialized workloads.

### Configuration Steps

When creating a new environment, users specify:

* Environment name
* Description
* Environment type
* Software version
* Programming language

### Spark Environment Example

If Spark is selected:

Additional options become available, including:

* Spark configuration
* Cluster settings

* Spark Type

After creation, the environment becomes available for notebook use.

---

## Switching Notebook Environments

A notebook can be reassigned to a different runtime environment.

### Required Steps

1. Stop the active kernel.
2. Open notebook environment settings.
3. Select a different environment.
4. Save the changes.
5. Restart the notebook.

### Environment Reinstantiation

Watson Studio automatically:

* Provisions the new runtime.
* Loads required libraries.
* Configures processing resources.

This allows users to move between Python, Scala, Spark, and other environments as project needs evolve.

---

## Updating Notebook Kernels

When a notebook switches environments, the original kernel may no longer match the new runtime.

### Kernel Update Process

Users may need to:

* Select a new kernel.
* Remove incompatible code cells.
* Reinsert data-loading code.
* Re-execute notebook cells.

This ensures compatibility with the selected environment.
