## What is Jupyter Notebook?

Jupyter Notebook is an interactive development environment that allows users to combine:

* Code
* Visualizations
* Documentation

within a single document.

This makes Jupyter Notebooks especially useful for:

* Data Science
* Machine Learning
* Data Analysis
* Research
* Education

A notebook can contain executable code, charts, tables, images, and explanatory text all in one place.

## Understanding Cells

Jupyter Notebooks are organized into cells.

A cell can contain:

* Code
* Markdown text

Each cell can be executed independently.

This allows users to develop and test code incrementally.

---

## Writing Python Code

A code cell can contain Python statements and expressions.

Example:

```python id="jp001"
x = 1
x
```

After executing the cell, Jupyter displays:

```python id="jp002"
1
```

Notice that the final expression is automatically displayed without requiring a `print()` statement.

---

## Working with Variables

Consider the following code:

```python id="jp003"
y = 2
y
```

Output:

```python id="jp004"
2
```

Jupyter automatically evaluates and displays the last expression in the cell.

This behavior is useful for quickly inspecting values.

---

## Running Cells

Code cells must be executed before their output becomes available.

A typical workflow is:

1. Write code.
2. Run the cell.
3. View the output.
4. Modify the code if necessary.
5. Run the cell again.

This interactive workflow makes experimentation easy.

---

## Using Libraries

Jupyter Notebooks work seamlessly with Python libraries.

Libraries provide pre-built functionality for:

* Numerical computing
* Data analysis
* Visualization
* Machine learning

One commonly used library is **NumPy**.

---

## Creating a NumPy Array

NumPy provides an efficient data structure called an array.

Example:

```python id="jp005"
import numpy as np

x = np.array([
    1, 2, 3, 3, 3,
    4, 5, 6, 7, 8,
    9, 1, 1, 2, 3,
    4, 5
])
```

The variable `x` now contains a NumPy array.

---

## Inspecting Data Types

To check the type of an object:

```python id="jp006"
type(x)
```

Example output:

```python id="jp007"
numpy.ndarray
```

This indicates that `x` is a NumPy array.

---

## Data Visualization

One of the strengths of Jupyter Notebooks is the ability to display visualizations directly within the notebook.

Visualization libraries can generate charts and graphs immediately after execution.

This makes exploratory data analysis more interactive.

---

## Using Seaborn

**Seaborn** is a Python visualization library built on top of Matplotlib.

Advantages of Seaborn include:

* Simpler syntax
* Attractive default styles
* Statistical plotting capabilities
* Better visual appearance

Import Seaborn:

```python id="jp008"
import seaborn as sns
```

It is common practice to place all import statements near the top of the notebook.

---

## Creating a Histogram

A histogram shows the distribution of values within a dataset.

Using Seaborn:

```python id="jp009"
sns.distplot(x)
```

The notebook displays a histogram directly below the cell.

The chart helps visualize:

* Frequency distribution
* Data concentration
* Potential patterns

---

## Seaborn and Matplotlib

Seaborn is built on top of **Matplotlib**.

Relationship:

```text id="jp010"
Seaborn
    ↓
Matplotlib
```

Matplotlib provides the underlying plotting functionality, while Seaborn improves the visual presentation and simplifies common plotting tasks.

---

## Documentation in Jupyter Notebooks

In addition to code cells, Jupyter supports documentation cells.

Documentation cells use **Markdown** formatting.

Markdown allows users to create:

* Titles
* Headings
* Lists
* Links
* Emphasis
* Images

This makes notebooks easier to understand and share.

---

## Creating a Markdown Cell

A cell can be changed from:

```text id="jp011"
Code
```

to:

```text id="jp012"
Markdown
```

using the notebook interface.

Once converted, the cell can contain formatted text.

---

## Markdown Example

Example Markdown content:

```markdown id="jp013"
# This is a Title

This is interesting.

## This is even more interesting.
```

When rendered, the notebook displays formatted headings and text.

---

## Benefits of Markdown

Markdown allows users to document:

* Project objectives
* Analysis steps
* Results
* Observations
* Conclusions

directly alongside executable code.

This improves readability and reproducibility.

---

## Combining Code, Visualizations, and Documentation

A key advantage of Jupyter Notebooks is the ability to combine multiple elements in a single document.

Example workflow:

```text id="jp014"
Markdown Explanation
        ↓
Python Code
        ↓
Visualization
        ↓
Analysis Notes
```

This structure helps create interactive reports and reproducible research documents.
