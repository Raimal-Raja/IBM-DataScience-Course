## What is JupyterLab?

JupyterLab is an advanced interactive development environment built on top of the Jupyter ecosystem.

It extends the capabilities of traditional Jupyter Notebooks by providing:

* Multiple document support
* File management tools
* Terminal access
* Flexible layouts
* Interactive workspaces

JupyterLab is designed to help users manage complex data science workflows more efficiently.

## Kernel Sessions

A key concept in JupyterLab is the **kernel session**.

A kernel is responsible for executing code contained within a notebook.

Relationship:

```text id="jl001"
Notebook
    ↓
Kernel Session
    ↓
Programming Language Runtime
```

For a Python notebook, the kernel typically contains a Python interpreter.

---

## How Kernels Work

When code is executed:

1. The notebook sends code to the kernel.
2. The kernel processes the code.
3. Results are returned to the notebook.
4. Output is displayed in the browser.

This architecture separates the notebook interface from the execution environment.

---

## Managing Kernel Sessions

JupyterLab allows users to:

* View active kernels
* Restart kernels
* Shut down kernels
* Monitor running sessions

This is especially useful when working with multiple notebooks simultaneously.

---

## Working with Multiple Tabs

One of JupyterLab's major advantages is support for multiple tabs.

Users can open:

* Several notebooks
* Multiple terminals
* Text files
* Data files

at the same time.

Example:

```text id="jl002"
Notebook 1
Notebook 2
Terminal 1
Terminal 2
```

## Executing Shell Commands

Jupyter notebooks can execute shell commands directly.

A shell command is prefixed with an exclamation mark (`!`).

Example:

```python id="jl005"
!pip install package_name
```

The command is executed by the operating system rather than the Python interpreter.

---

## Installing Packages

A common practice is installing or updating dependencies at the beginning of a notebook.

Example:

```python id="jl006"
!pip install --upgrade package_name
```

Benefits include:

* Consistent environments
* Reproducible notebooks
* Automatic dependency management

This ensures that required packages are available before executing notebook code.

---

## Cell Execution Indicators

When a notebook cell is running, Jupyter displays an indicator.

Example:

```text id="jl007"
[*]
```

The asterisk indicates that the cell is currently being executed.

After completion, a number appears.

Example:

```text id="jl008"
[1]
```

The number represents the execution order of the cell.

---

## Execution Counts

Each executed cell receives a unique execution number.

Example:

```text id="jl009"
[1]
[2]
[3]
```

These numbers help track the sequence in which notebook cells were run.

This is useful when troubleshooting notebook behavior.

---

## Checking Package Versions

Many notebooks verify that the correct package versions are installed.

Example:

```python id="jl010"
import tensorflow as tf

print(tf.__version__)
```

Version checks help ensure compatibility between code and dependencies.

This is especially important for machine learning libraries.

## Long-Running Tasks

Sometimes a notebook cell may:

* Take too long to execute
* Become unresponsive
* Enter an infinite loop

In such cases, the kernel can be restarted.

---

## Restarting the Kernel

Restarting the kernel creates a fresh execution environment.

Benefits include:

* Clearing memory
* Stopping running code
* Resetting variables
* Recovering from errors

After a restart:

```text id="jl011"
Variables → Removed
Memory → Cleared
Kernel → Restarted
```

All previously defined variables must be recreated by rerunning notebook cells.

---

## Restarting and Clearing Output

JupyterLab provides an option to:

```text id="jl012"
Restart Kernel and Clear All Outputs
```

This action:

1. Restarts the kernel.
2. Removes displayed outputs.
3. Creates a clean notebook state.

This is useful before sharing or exporting notebooks.
