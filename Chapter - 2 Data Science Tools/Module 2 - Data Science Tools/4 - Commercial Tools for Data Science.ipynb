### Commercial Tools for Data Science

#### Introduction
In previous lessons, we explored open source tools used in data science. In this section, we focus on commercial tools commonly used in enterprise environments and examine how they support different stages of the data science lifecycle.

---

### Commercial Data Management Tools

#### Enterprise Data Storage

Most enterprise organizations store critical business data using commercial database management systems.

#### Industry-Standard Platforms

The most widely used commercial databases include:

- Oracle Database
- Microsoft SQL Server
- IBM Db2

#### Why Commercial Databases Remain Popular

Although open source databases continue to grow in popularity, these commercial solutions remain dominant because:

- They are highly reliable
- They offer enterprise-grade support
- They provide extensive vendor ecosystems
- Organizations depend on them for mission-critical operations

#### Importance of Vendor Support

Data is one of the most valuable assets within an organization. Commercial vendors provide:

- Technical support
- Consulting services
- Partner ecosystems
- Long-term maintenance

These services are often a deciding factor for enterprise adoption.

---

### Commercial Data Integration and ETL Tools

#### What is ETL?

ETL stands for:

1. Extract
2. Transform
3. Load

ETL processes move and prepare data for analysis.

#### Market Leaders

According to industry evaluations such as the Gartner Magic Quadrant, leading ETL platforms include:

- Informatica PowerCenter
- IBM InfoSphere DataStage

#### Other Major Vendors

Additional ETL solutions are provided by:

- SAP
- Oracle
- SAS
- Talend
- Microsoft

#### Key Features

These tools typically provide:

- Graphical workflow design
- Automated pipeline deployment
- Data transformation capabilities
- Integration with commercial and open source systems

### Commercial Machine Learning and Data Mining Tools

#### Building Predictive Models

Commercial machine learning solutions often originate from the data mining industry.

#### Major Products

Popular commercial model-building platforms include:

- SPSS Modeler
- SAS Enterprise Miner

---

### Model Deployment in Commercial Platforms

#### Integrated Deployment

Commercial platforms tightly integrate model deployment into the model development workflow.

#### Example

SPSS Collaboration and Deployment Services provide deployment capabilities for assets created within the SPSS ecosystem.

#### Benefits

Integrated deployment enables organizations to:

- Publish models quickly
- Automate scoring processes
- Manage production assets
- Simplify operational workflows

---

### Open Model Formats

#### Predictive Model Markup Language (PMML)

Many commercial tools support exporting models using PMML.

#### Advantages

PMML enables:

- Model portability
- Interoperability
- Cross-platform deployment

#### Example

SPSS Modeler can export models in PMML format, allowing them to be consumed by both commercial and open source systems.

---

#### Current State of the Market

Model monitoring remains an emerging discipline.

Organizations often rely on open source monitoring tools because they provide:

- Faster innovation

### IBM InfoSphere Information Governance Catalog

#### Core Capabilities

This platform supports:

- Data dictionaries
- Metadata management
- Data stewardship
- Data lineage

#### Data Stewardship

Each data asset is assigned to a responsible owner called a **Data Steward**.

#### Responsibilities of Data Stewards

Data stewards:

- Maintain data quality
- Manage metadata
- Serve as points of contact
- Ensure compliance

---

### Data Lineage

#### What is Data Lineage?

Data lineage tracks the origin and transformation history of data assets.

#### Benefits

Users can:

- Trace data back to source systems
- Understand transformation processes
- Verify data quality
- Improve transparency
